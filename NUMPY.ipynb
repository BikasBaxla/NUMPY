```python
import time
n = 1000
lst = list(range(n))
start = time.time()
lst = [x * 2 for x in lst]
print("List Time :",time.time()-start)
```

```python
arr = np.arange(n)
start = time.time()
arr = arr *2
print("Numpy Time: ",time.time()-start)
```
